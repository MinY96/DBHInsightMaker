방법: 잔차 분포를 히스토그램 혹은 커널밀도추정(KDE)으로 그림.

체크 포인트: 정규분포 모양과 유사한지 확인.

3. Q-Q Plot (Quantile–Quantile Plot)

방법: 잔차의 분위수 vs 정규분포 분위수를 비교.

꼬리 부분이 위/아래로 크게 벗어나면 heavy-tail / light-tail 문제.

체크 포인트: 큰 벗어남이 없고 직선에 가까운지 확인.

체크 포인트: Cook’s distance 기준선을 벗어나는 점이 있는지.

🧾 종합적으로 얻을 수 있는 통찰

선형성(Linearity)

Residual vs Fitted에서 곡선 패턴이 없어야 함.

등분산성(Homoscedasticity)

Residual vs Fitted, Scale-Location Plot에서 잔차 분산이 일정해야 함.

독립성(Independence)

잔차가 특정 패턴 없이 무작위적 분포. (시계열 데이터는 Durbin-Watson 통계량 확인)

정규성(Normality)

잔차 히스토그램, Q-Q plot에서 정규분포와 유사해야 함.

이상치 및 영향점 확인

Residual vs Leverage, Cook’s distance로 영향이 큰 점을 식별.

✅ 통계적으로 유의미한 회귀분석으로 보려면?

계수의 p-value < 0.05: 개별 독립변수가 종속변수에 유의미하게 기여.

F-통계량의 p-value < 0.05: 모델 전체가 유의미함.

R², Adj R² 높음: 설명력 충분.

잔차 분석에서 가정 만족: 선형성·등분산성·정규성 위반이 없을 것.

이상치/영향점이 모델을 왜곡하지 않음: Cook’s distance 큰 점이 없거나 제거 후에도 결과가 유사.
